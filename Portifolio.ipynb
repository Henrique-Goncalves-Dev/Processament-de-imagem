```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow
```

```python
img_rgb = cv2.imread("Short-N-Sweet-Edition-Deluxe.jpg")
cv2_imshow(img_rgb)
img_rgb


```

```python
img_bgr = cv2.imread("Short-N-Sweet-Edition-Deluxe.jpg")
img_grey = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

img_grey
```

```python
# 1 bit = 2 níveis
niveis_1 = 2
img_1bit = np.round(img_grey / 255 * (niveis_1 - 1))
img_1bit = img_1bit * (255 / (niveis_1 - 1))

# 2 bits = 4 níveis
niveis_2 = 4
img_2bits = np.round(img_grey / 255 * (niveis_2 - 1))
img_2bits = img_2bits * (255 / (niveis_2 - 1))

# 4 bits = 16 níveis
niveis_4 = 16
img_4bits = np.round(img_grey / 255 * (niveis_4 - 1))
img_4bits = img_4bits * (255 / (niveis_4 - 1))

# 8 bits = 256 níveis
niveis_8 = 256
img_8bits = np.round(img_grey / 255 * (niveis_8 - 1))
img_8bits = img_8bits * (255 / (niveis_8 - 1))
```

```python
fig, ax = plt.subplots(1, 5, figsize=(15, 4))

ax[0].imshow(img_grey, cmap='gray')
ax[0].set_title("Original")
ax[0].axis('off')

ax[1].imshow(img_1bit, cmap='gray')
ax[1].set_title("1 bit")
ax[1].axis('off')

ax[2].imshow(img_2bits, cmap='gray')
ax[2].set_title("2 bits")
ax[2].axis('off')

ax[3].imshow(img_4bits, cmap='gray')
ax[3].set_title("4 bits")
ax[3].axis('off')

ax[4].imshow(img_8bits, cmap='gray')
ax[4].set_title("8 bits")
ax[4].axis('off')

plt.tight_layout()
plt.savefig('quantizacao.png')
plt.show()
```
