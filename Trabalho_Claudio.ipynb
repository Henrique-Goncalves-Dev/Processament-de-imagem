```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1 - ILUSÃO DE CONTRASTE
# ==========================================

# Cria duas imagens de 200x200
imagem1 = np.zeros((200, 200))
imagem2 = np.zeros((200, 200))

# Fundo escuro
imagem1[:, :] = 0.2

# Fundo claro
imagem2[:, :] = 0.8

# Quadrado central com o MESMO tom de cinza
imagem1[60:140, 60:140] = 0.5
imagem2[60:140, 60:140] = 0.5


plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(imagem1, cmap="gray", vmin=0, vmax=1)
plt.title("Fundo escuro")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(imagem2, cmap="gray", vmin=0, vmax=1)
plt.title("Fundo claro")
plt.axis("off")

plt.show()


# Mostra o valor dos dois quadrados
print("Valor do quadrado 1:", imagem1[100, 100])
print("Valor do quadrado 2:", imagem2[100, 100])
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Cria uma imagem
imagem = np.zeros((200, 600))

# Cada pedaço recebe um tom de cinza diferente

imagem[:, 0:100] = 40
imagem[:, 100:200] = 80
imagem[:, 200:300] = 120
imagem[:, 300:400] = 160
imagem[:, 400:500] = 200
imagem[:, 500:600] = 240


plt.figure(figsize=(10, 4))

plt.imshow(imagem, cmap="gray", vmin=0, vmax=255)

plt.title("Bandas de Mach")

plt.axis("off")

plt.show()
```

```python
linha = imagem[100]

plt.plot(linha)

plt.title("Intensidade dos pixels")

plt.xlabel("Posição do pixel")

plt.ylabel("Intensidade")

plt.show()
```

```python
bordas = np.zeros((200, 600))

for x in range(599):

    diferenca = imagem[100, x + 1] - imagem[100, x]

    if diferenca != 0:

        bordas[:, x] = 255


plt.imshow(bordas, cmap="gray")

plt.title("Bordas encontradas")

plt.axis("off")

plt.show()
```
