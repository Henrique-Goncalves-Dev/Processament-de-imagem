```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
img_rgb = cv2.imread("vox.jpg")
img_rgb
```

```python
img_bgr = cv2.imread("vox.jpg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
img_grey = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

img_grey
```

```python
plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("RGB")

plt.subplot(1,2,2)
plt.imshow(img_bgr)
plt.title("BGR")

plt.show()


```

```python
fig, ax =   plt.subplots(2,2, figsize=(6,4))

#1a imagem
ax[0,0].imshow(img_rgb)
ax[0,0].set_title("RGB")
ax[0,0].axis('off')
#2a imagem
ax[0,1].imshow(img_bgr)
ax[0,1].set_title("BGR")
ax[0,1].axis('off')
#3a imagem
ax[1,0].imshow(img_hsv)
ax[1,0].set_title("HSV")
ax[1,0].axis('off')

#4a imagem
ax[1,1].imshow(img_grey, cmap = 'gray')
ax[1,1].set_title("Cinza")
ax[1,1].axis('off')


plt.tight_layout()
plt.savefig('conversao_de_cores.png')
plt.show()
```
